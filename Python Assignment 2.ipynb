```python
### 1. Storing message in variable - upper, lower, title case
message = 'I love python. It is Very easy to code'
print("Message in upper case - ", message.upper())
print("Message in lower case - ", message.lower())
print("Message in title case - ", message.title())
print("Message in capitalize case - ", message.capitalize())
```

```python
### 9.	Ask the user for two numbers and print Sum, Difference, Product, Quotient
n1 = int(input("Please enter the first number: "))
n2 = int(input("Please enter the second number: "))
print("Sum : ", n1+n2)
print("Difference : ", n1-n2)
print("Product : ", n1*n2)
print("Quotient : ", n1//n2)
```

```python
### 12.	Ask the user to enter a number and print True if it is greater than 50, otherwise print False
n3 = int(input("Please enter the number: "))
greater = lambda x: "True" if x > 50 else "False"
print("Is number greater than 50: ", greater(n3))
```

```python
### 13.	Ask the user to enter a number and print True if it is even, otherwise print False.
n4 = int(input("Please enter the number: "))
even_number = lambda x: "True" if x%2==0 else "False"
print("Is this number even? ", even_number(n4))
```

```python
### 14.	Ask the user to enter their age and check if they are eligible to vote (age ≥ 18). Print True or False.
voter_age = int(input("Please enter the age: "))
vote_eligible = lambda x: "True" if x >= 18 else "False"
print("Eligilble to Vote? ", vote_eligible(voter_age))
```

```python
### 15.	Ask the user to enter two numbers and print True if the first number is greater than the second, else print False.
first_num = int(input("Please enter the first number: "))
second_num = int(input("Please enter the second number: "))
check = lambda a, b: "True" if a > b else "False"
print("Is first number greater than second number? ", check(first_num, second_num))
```

```python
### 16. Ask the user to enter a word and check if its length is more than 5 characters. Print True or False.
word = input("Please enter the word: ")
len_check = lambda a: "True" if len(word) > 5 else "False"
print("Is word length greater than 5? ", len_check(word))
```
