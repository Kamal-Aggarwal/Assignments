### Step 9. Which was the most-ordered item?

```python
print("The most-ordered item is - ", chipo['item_name'].value_counts().idxmax())
```

### Step 10. For the most-ordered item, how many items were ordered?

```python
item_count = chipo['item_name'].value_counts()
most_order_item = item_count.idxmax()
most_order_count = item_count.values[0]
print(f"Most ordered item - {most_order_item} was ordered {most_order_count} times")
```

### Step 13. Turn the item price into a float

#### Step 13.a. Check the item price type

```python
print("Item price datatype - ", chipo["item_price"].dtype)
```

#### Step 13.b. Create a lambda function and change the type of item price

```python
chipo["item_price"] = pd.to_numeric(chipo["item_price"].replace('[$,]','',regex = True))
print("New Data type of item price - ", chipo["item_price"].dtypes)
```

### Step 16. What is the average revenue amount per order?

```python
# Solution 1
Tot_Revenue = chipo["item_price"].sum()
Tot_Order_Count = chipo['order_id'].nunique()
Avg_Revenue = round(Tot_Revenue / Tot_Order_Count, 2)
print("Average Revenue Amount per Order - ", Avg_Revenue)
```
