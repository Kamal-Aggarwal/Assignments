```python
### 21. Dictionary from Two Lists:
keys = [1, 2, 3, 4, 5]
values = ['Ram', 'Sham', 'John', 'Peter', 'Sam']
roll_name = dict(zip(values, keys))
print(roll_name)
```

```python
### 22. Word Length Mapping:
keys = ['Ram', 'Sham', 'John', 'Peter', 'Sam']
word_lengths = {word: len(word) for word in keys}
print(word_lengths)
```

```python
### 23. Prime Numbers in a Range
primes = [x for x in range(2, 101) if all(x % y != 0 for y in range(2, x))]

# Print the resulting list of primes
print(primes)
```

```python
### 24. Flatten a Nested List
list2 = [[1,2],[3,4],[5,6,7]]
list3 = [item for sublist in list2 for item in sublist]
print(list3)
```

```python
### 25. Character Frequency Counter:
input_str = "ILovePythonProgramming"
## Converting string to lower case 
lower_str = input_str.lower()
letter_len = {char: lower_str.count(char) for char in lower_str}
print(letter_len) 
```
