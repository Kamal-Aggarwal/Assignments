```python
### 16. List Square Generator:
squares = [x**2 for x in range(1, 21)]
print(squares)
```

```python
### 17. Even and Odd Filter:
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
# List comprehension for even numbers
even_numbers = [num for num in numbers if num % 2 == 0]
# List comprehension for odd numbers
odd_numbers = [num for num in numbers if num % 2 != 0]
print("Even number: ", even_numbers)
print("Odd number: ", odd_numbers)
```

```python
### 18. Vowel Counter:
str1 = input("Please enter the string: ")
count = 0
## list of vowels
vowels = ["a", "e", "i", "o", "u"]
for i in range(len(str1)):
    if str1[i] in vowels:
        count+= 1
print("Vowel count in the string is: ", count)
    
```

```python
### 19. Multiplication Table:
number = int(input("Enter enter the number: "))
for i in range(1, 11):
    print(f"{number} * {i} = ", number * i)
```

```python
### 20. Common Elements Finder:
list1 = [1, 4, 6, 7, 'aB', 'b', 'c']
list2 = [2, 6, 1, 'c', 'ab', 'd']
common_elements = [a for a in list1 if a in list2]
print(common_elements)
```
